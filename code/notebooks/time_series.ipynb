```python
data_dir='../../output/mongodb/'
```

```python
from plotly.subplots import make_subplots

json_files = [f for f in os.listdir(data_dir) if f.endswith('.json') and 'top5_countries_' in f]

# Create subplot grid: 2 rows, 5 columns
fig = make_subplots(rows=2, cols=5, shared_xaxes=False, shared_yaxes=False,
                    subplot_titles=[f.split('_')[3] for f in json_files[:10]])

for i, filename in enumerate(json_files[:10]):
    df = pd.read_json(os.path.join(data_dir, filename))
    country_code = df['countriesAndTerritories'].iloc[0].replace("_", " ")
    row = i // 5 + 1
    col = i % 5 + 1
    if 'cases' in filename:
        fig.add_trace(
            go.Scatter(x=df['dateRep'], y=df['cases'], mode='lines', name=country_code),
            row=row, col=col
        )
    else:
        fig.add_trace(
            go.Scatter(x=df['dateRep'], y=df['deaths'], mode='lines', name=country_code),
            row=row, col=col
        )
    # -- remove x ticks
    fig.update_xaxes(showticklabels=False, row=row, col=col)

fig.update_layout(height=800, width=1400, title_text="Top 5 Countries - Daily Deaths Over Time")
fig.show()
```

```python
group1 = json_files[:5]
group2 = json_files[5:10]

def build_figure(file_group, title):
    fig = make_subplots(rows=5, cols=1, shared_xaxes=False, shared_yaxes=False,
                        subplot_titles=[f.split('_')[3] for f in file_group])

    for i, filename in enumerate(file_group):
        df = pd.read_json(os.path.join(data_dir, filename))
        country_code = df['countriesAndTerritories'].iloc[0].replace("_", " ")
        y_col = 'cases' if 'cases' in filename else 'deaths'
        fig.add_trace(
            go.Scatter(x=df['dateRep'], y=df[y_col], mode='lines', name=country_code),
            row=i+1, col=1
        )
        fig.update_xaxes(showticklabels=False, row=i+1, col=1)

    fig.update_layout(height=1200, width=600, title_text=title)
    return fig

# Create and show both figures
fig1 = build_figure(group1, "top 5 countries: cases")
fig2 = build_figure(group2, "top 5 countries: deaths")

fig1.show()
fig2.show()
```
