```python
import os
import plotly.express as px
import plotly.graph_objects as go

import pandas as pd
import pycountry
os.getcwd()

data_dir='../../output/'

def get_iso_codes(name):
    '''get iso codes for map visualization'''
    try:
        name = name.replace("_", " ")
        country = pycountry.countries.search_fuzzy(name)[0]
        return pd.Series({"ISO2": country.alpha_2, "ISO3": country.alpha_3})
    except:
        return pd.Series({"ISO2": None, "ISO3": None})
```

```python
def plot_total_deaths_per_country(df):

    df[["ISO2", "ISO3"]] = df["_id"].apply(get_iso_codes)
    
    fig = go.Figure(go.Choropleth(
        locations=df["ISO3"],
        z=df["logTotalDeaths"],
        text=df["_id"], 
        hovertemplate='<b>%{text}</b><br>Deaths: %{customdata[0]}<extra></extra>',
        customdata=df[["totalDeaths"]],
        colorscale="Plasma",
        colorbar_title="log(Deaths)"
    ))

    fig.update_layout(
        title_text="Total COVID-19 Deaths by Country",
        margin=dict(l=0, r=0, t=50, b=0)
    )

    fig.update_geos(
        showcoastlines=True,
        showland=True,
        showframe=False,
        fitbounds="locations"             
    )

    fig.show()
    return fig

def plot_total_cases_per_country(df):

    df[["ISO2", "ISO3"]] = df["_id"].apply(get_iso_codes)
    
    fig = go.Figure(go.Choropleth(
        locations=df["ISO3"],
        z=df["logTotalCases"],
        text=df["_id"], 
        hovertemplate='<b>%{text}</b><br>Cases: %{customdata[0]}<extra></extra>',
        customdata=df[["totalCases"]],
        colorscale="viridis",
        colorbar_title="log(Cases)"
    ))

    fig.update_layout(
        title_text="Total COVID-19 Cases by Country",
        margin=dict(l=0, r=0, t=50, b=0)
    )

    fig.update_geos(
        showcoastlines=True,
        showland=True,
        showframe=False,
        fitbounds="locations"             
    )

    fig.show()
    return fig
```

## MongoDB

```python
mongo_file=data_dir+'mongodb/country_total_stats.json'
output='../../figures/mongodb/'
df=pd.read_json(mongo_file)
```

```python
mongo_total_deaths_fig=plot_total_deaths_per_country(df)
mongo_total_deaths_fig.write_html(output+'mongodb_total_deaths_per_country.html')
```

```python
mongo_total_cases_fig=plot_total_cases_per_country(df)
mongo_total_cases_fig.write_html(output+'mongodb_total_cases_per_country.html')
```
